```python
# %pip install numpy pandas scikit-learn

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn import svm
from sklearn.metrics import accuracy_score
```

Data collection and Analysis


```python
#loading the data from csv file to a pandas Dataframe
# loading the data from csv file to a pandas Dataframe
parkinsons_data = pd.read_csv('Parkinsson_disease.csv')

```

```python
#grouping the data based on the target variable
parkinsons_data.groupby('status').mean(numeric_only=True)
```

Data preprocessing


separating the features and target

```python
X = parkinsons_data.drop(columns=['name', 'status'])   #when you drop a column, you need to specify the axis. axis=0 means you are dropping rows, axis=1 means you are dropping columns.
Y = parkinsons_data['status']  # this is target variable, which we want to predict. It is a binary variable, 1 means the person has Parkinson's disease, 0 means the person does not have Parkinson's disease.
```

```python
print(Y)
```

Splitting the data into training data and testing data

```python
X_train , X_test, Y_train  , Y_test = train_test_split(X,Y,test_size=0.2, random_state = 42)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Data standardisation


```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
```

```python
X_train = scaler.transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
print(X_train)
```

Model Training

Support Vector Machine

```python
model = svm.SVC(kernel='linear')
```

```python
#training the SVM model with training data
model.fit(X_train, Y_train)
```

Model Evaluation

```python
# accuracy score on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
```

```python
print('Accuracy score of training data: ', training_data_accuracy)
```

```python
# accuracy score on training data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
```

```python
print('Accuracy score of test data: ', test_data_accuracy)
```

Building a Predictive System


```python
input_data = (119.992, 157.302, 74.997, 0.00784, 0.00007, 0.0037, 0.00554, 0.01109, 0.04374, 0.426, 0.02182, 0.0313, 0.02971, 0.06545, 0.02211, 21.033, 0.414783, 0.815285, -4.813031, 0.266482, 2.301442, 0.284654)

# changing input data to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the numpy array
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

# standardize the data
std_data = scaler.transform(input_data_reshaped)

prediction = model.predict(std_data)
print("Prediction:", prediction)

if prediction[0] == 0:
    print("The person does not have Parkinson's Disease")
else:
    print("The person has Parkinson's Disease")
```

### Export / Save Model and Scaler

```python
import joblib

# Export / Save trained SVM model and scaler for deployment and comparison
joblib.dump(model, 'parkinsons_svm_model.pkl')
joblib.dump(scaler, 'parkinsons_scaler.pkl')
print("SVM model and scaler successfully saved to 'parkinsons_svm_model.pkl' and 'parkinsons_scaler.pkl'!")
```
